문자 선택 연산자(인덱싱): []

```python
# 문자 시작은 0부터 시작함~
print("문자 선택 연산자에 대해 알아보자")
print("안녕하세요"[0])
print("안녕하세요"[1])
print("안녕하세요"[2])
print("안녕하세요"[3])
print("안녕하세요"[4])
print("")
print("문자 선택 연산자를 거꾸로 출력해보자")
print("안녕하세요"[-1])
print("안녕하세요"[-2])
print("안녕하세요"[-3])
print("안녕하세요"[-4])
print("안녕하세요"[-5])
```

문자열 범위 선택 연산자(슬라이싱): [:] (범위)

```python
print("012345"[1:3]) # 1부터 3미만까지 출력해라
print("012345"[1:]) # 1부터 끝까지 출력해라
print("012345"[:5]) # 5미만까지 출력해라
```

input() 함수는 문자를 입력받던, 숫자를 입력받던 항상 결과는 문자열 자료형으로 출력됨

```python
#int() 함수 활용하기
string_a = input("입력A>")
int_a = int(string_a) #문자열을 숫자로 받는 방법 int()를 사용하자

string_b = input("입력B>")
int_b = int(string_b)

print("문자열 자료:", string_a + string_b) #이건 숫자를 문자그대로 받아서 그냥 받은 숫자 나란히 출력됨
print("숫자 자료:", int_a + int_b)#숫자를 받았기때문에 연산됨
```

```python
#float() 함수 활용하기
string_a = input("입력A>")
float_a = float(string_a)

string_b = input("입력B>")
float_b = float(string_b)

print("문자열 자료:", string_a + string_b) #이건 숫자를 문자그대로 받아서 그냥 받은 숫자 나란히 출력됨
print("숫자 자료:", float_a + float_b)#숫자를 받았기때문에 연산됨
```

```python
#str() 함수 활용하기 >> 숫자를 문자열로 변환하기
output_a = str(52)
output_b = str(52.123)

print(type(output_a), output_a)
print(type(output_b), output_b)
```

```python
#int()함수와 float()함수 조합하기
input_a = float(input("첫 번째 숫자"))
input_b = float(input("두 번째 숫자"))

print("덧셈 결과:", input_a + input_b)
print("뺄셈 결과:", input_a - input_b)
print("곱셈 결과:", input_a * input_b)
print("나눗셈 결과:", input_a / input_b)
```

```python
#inch단위를 cm단위로 변경하기
change_input = input("inch 단위의 숫자를 입력해주세요: ") # 인치 입력받을 때 문구

inch = int(change_input) # 인치 입력받기
cm = inch * 2.54 # 정보 : 1인치는 2.54cm 이다.

print(inch, "inch는 cm 단위로", cm, "cm입니다.")
```

```python
#문자열 내 단어 변경해서 출력하기
a = input("문자열 입력>")
b = input("문자열 입력>")

swap = a
a = b
b = swap 

print(a, b)
```

```python
# format() 함수로 숫자를 문자열로 변환하기
# 형태 : "{}".format(매개변수)
string_a = "{}".format(12345)

# 출력하기
print(string_a)
print(type(string_a))
```

```python
# format() 함수로 숫자를 문자열로 변환하기
# 앞 괄호의 개수와 format 뒤 매개변수의 개수가 맞지 않으면 IndexError 노출됨
format_a = "{}만 원".format(5000)
format_b = "파이썬 열공하여 첫 연봉 {}만 원 만들기".format(5000)
format_c = "{}{}{}".format(3000, 4000, 5000)
format_d = "{}{}{}".format(1, "문자열", True)

print(format_a)
print(format_b)
print(format_c)
print(format_d)
```

```python
#정수를 특정칸에 출력하기
output_a = "{:d}".format(52) #기존대로 출력
output_b = "{:5d}".format(52) #총 5칸에 매개변수 뒤에서부터 출력 (앞에는 다 공백)
output_c = "{:10d}".format(52) #총 10칸에 매개변수 뒤에서부터 출력 (앞에는 다 공백)
output_d = "{:05d}".format(52) #output_b와 같은데 공백은 0으로 채움
output_e = "{:05d}".format(-52) #음수이기떄문에 부호,공백,매개변수 포함에서 5칸에 들어가야함


print(output_a)
print(output_b)
print(output_c)
print(output_d)
print(output_e)
```

```python
output_h = "{:+5d}".format(52) #기호랑 같이 뒤로가기
output_i = "{:=+5d}".format(52) #기호만 앞으로 밀어버리기
output_j = "{:+05d}".format(52) #기호랑 같이 0으로 채우기
# +대신  -로 적용해도 다 동일하다 ~


print(output_h)
print(output_i)
print(output_j)


print(len(output_h))
print(len(output_i))
print(len(output_j))
```

```python
#의미 없는 소수점 제거하기
output_a = 52.0
output_b = "{:g}".format(output_a)

print(output_a)
print(output_b)
```
