[문제1]  
리스트에 저장된 문자열 중에서 길이가 5인 것들만 리스트 형태로 출력하도록 구현하세요.  

a= ['alpha', 'bravo', 'charlie', 'delta', 'echo', 'foxtrot', 'golf', 'hotel', 'india']  

[출력결과]     
['alpha', 'bravo', 'delta', 'hotel', 'india']

```python
a= ['alpha', 'bravo', 'charlie', 'delta', 'echo', 'foxtrot', 'golf', 'hotel', 'india']  

result = [] # 빈 result 리스트 생성하고

for word in a:           # a 리스트의 단어를 하나씩 꺼내고(단어 끝까지 반복)
    if len(word) == 5:   # a 리스트 안 단어의 길이가 5인지 확인 후
        result.append(word) # 맞으면 빈 result 리스트에 추가

print(result)
```

[문제8]  

리스트에서 int만 추출하는 프로그램을 구현하세요.  
oldlist = [1, 2, 'A', False, 3]

[출력결과]   
[1, 4, 9]  


```python
oldlist = [1, 2, 'A', False, 3]

result = []

for x in oldlist:
    if isinstance(x, int) and not isinstance(x, bool): # 이건 조금 더 생각이 필요할듯
        result.append(x)
print(result)
```

[문제9]   
1부터 10까지의 홀수, 짝수로 출력하는 프로그램을 구현하세요.

[출력결과]   
{1: '홀수',
 2: '짝수',
 3: '홀수',
 4: '짝수',
 5: '홀수',
 6: '짝수',
 7: '홀수',
 8: '짝수',
 9: '홀수',
 10: '짝수'}


```python
result = {} #빈 result 딕셔너리 생성 딕셔너리는 {키:값} 형태

for i in range(1, 11): #범위는 1부터 11미만까지
    if i % 2 == 1: #i값을 2로 나눴을 때 나머지가 1이면
      result[i] = ': 홀수'
    else : #i값을 2로나눴을 때 나머지가 1이 아니면(즉, 나머지가 0)
      result[i] = ': 짝수' 

print(result)



```

 [문제10]  
 if~else문을 이용해서 년도에 해당하는 띠를 구하는 프로그램을 구현하시오.  
 단, 년도은 input()으로 입력받음.

   1. 12가지 띠  
      원숭이, 닭, 개, 돼지, 쥐, 소, 호랑이, 토끼, 용, 뱀, 말, 양  
   
   2. 띠구하는 공식  
     year%12의 결과가  
     0이면 "원숭이", 1이면 "닭"  2이면 "개",    3이면 "돼지",  
     4이면 "쥐",    5이면 "소", 6이면 "호랑이", 7이면 "토끼",  
     8이면 "용",    9이면 "뱀", 10이면 "말",   11이면 "양"  
     
   3. 출력결과  
     2012년도의 태생은  용띠입니다.  

```python
year = int(input('년도:'))

if year % 12 == 0:
    print(year,'년도의 태생은 원숭이띠입니다.')
elif year % 12 == 1:
    print(year,'년도의 태생은 닭띠입니다.')
elif year % 12 == 2:
    print(year,'년도의 태생은 개띠입니다.')
elif year % 12 == 3:
    print(year,'년도의 태생은 돼지띠입니다.')
elif year % 12 == 4:
    print(year,'년도의 태생은 쥐띠입니다.')
elif year % 12 == 5:
    print(year,'년도의 태생은 소띠입니다.')
elif year % 12 == 6:
    print(year,'년도의 태생은 호랑이띠입니다.')
elif year % 12 == 7:
    print(year,'년도의 태생은 토끼띠입니다.')
elif year % 12 == 8:
    print(year,'년도의 태생은 용띠입니다.')
elif year % 12 == 9:
    print(year,'년도의 태생은 뱀띠입니다.')
elif year % 12 == 10:
    print(year,'년도의 태생은 말띠입니다.')
else:
    print(year,'년도의 태생은 양띠입니다.')
```

[문제11]
list comprehension을 이용해서 1부터 10까지 합계 구하기

```python
result = [] #리스트 새로 만들고
for x in range(1, 11): #범위는 1(포함)부터 11미만까지
    result.append(x) #x의 값(1~10)을 result리스트에 추가
 
print(sum(result)) #result리스트에 있는 값 모두 더해서 출력
```

[문제12]
A 학급에 총 10명의 학생이 있다. 이 학생들의 중간고사 점수는 다음과 같다.

[70, 60, 55, 75, 95, 90, 80, 80, 85, 100]

list comprehension을 사용하여 A 학급의 평균 점수를 구해 보자.

```python
score = [70, 60, 55, 75, 95, 90, 80, 80, 85, 100]

# avg = sum(score) / len(score) << 제미나이의 추천방법이였음
# 근데 제미나이가 굳이 리스트 컴프리헨션을 사용한다고?라는식으로 말하긴했음

total = sum([i for i in score])
count = len([1 for _ in score])

avg = total / count
print(avg)
```
