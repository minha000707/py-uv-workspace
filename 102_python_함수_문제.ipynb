[문제2]   
문자열을 입력받아 각 문자들로 구성된 리스트로 반환하는 make_list 함수를 정의하세요.  
make_list("abcd")  

[출력결과]

['a', 'b', 'c', 'd']


```python
def make_list(string):
    result = []
    for i in string:
        result.append(i)
    return result

make_list("abcd")
```

[문제3]   
숫자로 구성된 하나의 리스트를 입력받아, 짝수들을 추출하여 리스트로 반환하는 pickup_even 함수를 구현하세요.   

pickup_even([3, 4, 5, 6, 7, 8])

[출력결과]   
[4, 6, 8]

```python
def pickup_even(item):
    result = [] # 결과값 담을 result 리스트 생성
    for i in item: # 리스트 내 요소 다 확인할때까지 반복
        if i % 2 == 0: # 조건 : 리스트 안의 요소들이 짝수일 때
            result.append(i) # result리스트에 추출한 i값(짝수) 추가
    return result # 최종 결과값 출력

pickup_even([3, 4, 5, 6, 7, 8])
```

[문제4]  
세 개의 숫자를 입력받아 가장 큰수를 출력하는 print_max 함수를 정의하라. 단 if 문을 사용해서 수를 비교하세요.   
max(10, 2, 5)

[출력결과]  
10

```python
def max(a, b, c):
    if a > b and a > c: # a가 b와 c보다 크면
        return a # a출력
    elif b > a and b > c:# b가 a와 c보다 크면
        return b # b출력
    else: # 그외는 
        return c #c 출력
    
max(10, 2, 5)
```
