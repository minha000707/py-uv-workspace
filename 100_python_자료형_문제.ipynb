문제3)   
'spam ham'을 'ham spam'으로 치환하세요.   

※문자열의 split( )과 join( )메서드 그리고 리스트의 reverse( )메서드를 사용

```python
a = 'spam ham'
a = a.split() #쪼개고
a.reverse() #순서바꾸고

a = ' '.join(a) #리스트에 있는 것들을 꺼내서 공백으로 연결
print(a)
```

문제14)

```python
l = [1,3,5,4,2]
l.sort() # 리스트 오름차순으로 정렬
l.reverse() # 리스트 역순으로 배열
print(l)

#dir(list) list의 내장함수 검색해봄
```
