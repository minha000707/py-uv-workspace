formatting 방법은 3가지가 있음

1) % 포매팅
2) format() 메서드
3) f-string(가장 현대적이고 추천하는 방식)
=> 가독성, 간결성, 성능 모두 완벽하게 개선

```python
# 035번 (% formatting의 개념을 잘 모름)
name1 = "김민수"
age1 = 10
name2 = "이철희"
age2 = 13

print("이름:", name1 , "나이:", age1)
print("이름:", name2 , "나이:", age2) #나의 정답


name1 = "김민수" 
age1 = 10
name2 = "이철희"
age2 = 13
print("이름: %s 나이: %d" % (name1, age1))
print("이름: %s 나이: %d" % (name2, age2)) #wikidocs 정답

# %s = 문자열
# %d = 정수 digit
# %f = 실수
```

```python
# 036번
name1 = "김민수"
age1 = 10
name2 = "이철희"
age2 = 13

print("이름:{} 나이:{}".format(name1, age1))
print("이름:{} 나이:{}".format(name2, age2))
```
