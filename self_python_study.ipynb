```python
# 086번 ⭐⭐⭐
# append를 생각했었으나 append는 튜플에서 사용이였음
# 딕셔너리에 접근 또는 데이터를 추가할때는 [] 대괄호를 이용

ice = {"메로나": 1000, "폴라포": 1200, "빵빠레": 1800}
ice["죠스바"] = 1200
ice["월드콘"] = 1500
print(ice)

```

```python
# 086번 꼬리질문
# #한번에 여러개는 못 추가하는지? 
ice = {"메로나": 1000, "폴라포": 1200, "빵빠레": 1800}
ice["죠스바", "월드콘"] = 1200, 1500

print(ice)
```

```python
# 088번
ice = {'메로나': 1000,
       '폴로포': 1200,
       '빵빠레': 1800,
       '죠스바': 1200,
       '월드콘': 1500}

ice["메로나"] = 1300

print(ice)
```

```python
# 089번 ⭐⭐ 딕셔너리에서 데이터를 삭제할때는 del 을 사용
ice = {'메로나': 1000,
       '폴로포': 1200,
       '빵빠레': 1800,
       '죠스바': 1200,
       '월드콘': 1500}

del ice["메로나"]

print(ice)
```
